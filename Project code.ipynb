```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Make plots look nicer
sns.set(style="whitegrid")
%matplotlib inline
```

```python
df = pd.read_csv("online food delivery dataset.csv")
df.head()
```

```python
print("Shape of dataset (rows, columns):", df.shape)
print("\nColumn names:\n", df.columns.tolist())
print("\nData types:\n")
df.info()
```

```python
df.describe(include="all")
```

```python
null_counts = df.isnull().sum()
print("Null values in each column:\n")
print(null_counts)

total_nulls = null_counts.sum()
print("\nTotal null values in the dataset:", total_nulls)
```

```python
if total_nulls > 0:
    for col in df.columns:
        if df[col].isnull().sum() > 0:
            if pd.api.types.is_numeric_dtype(df[col]):
                mean_value = df[col].mean()
                df[col] = df[col].fillna(mean_value)
                print(f"Filled null values in numeric column '{col}' with mean = {mean_value:.2f}")
            else:
                mode_value = df[col].mode()[0]
                df[col] = df[col].fillna(mode_value)
                print(f"Filled null values in categorical column '{col}' with mode = '{mode_value}'")
else:
    print("No null values found in the dataset. Nothing to fill.")
```

```python
print("Null values after cleaning:\n")
print(df.isnull().sum())
print("\nTotal remaining null values:", df.isnull().sum().sum())
```
